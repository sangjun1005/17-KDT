````python
import pandas as pd
import numpy as np

# 데이터 로드
data_url = "https://raw.githubusercontent.com/YoungjinBD/dataset/main/airquality.csv"
airquality = pd.read_csv(data_url)
airquality.head()
````

## 문제 ① 결측치 처리 및 평균 비교

`airquality` 데이터에 대해서 **결측치가 가장 많은 변수**를 찾아서 해당 결측치를 **0으로 대치**하고,  
**결측치를 제외한 평균**과 **0으로 대치한 후의 평균**과의 차이를 구하시오.

> 출력: 두 평균의 차이 (절대값 여부는 풀이 방식에 따름)

````python
missing_col = airquality.isnull().sum().idxmax()
mean_before = airquality[missing_col].mean()
mean_after = airquality[missing_col].fillna(0).mean()
diff = abs(mean_before - mean_after)

print("결측치가 가장 많은 변수:", missing_col)
print("결측치 제외 평균:", round(mean_before, 4))
print("0 대치 후 평균:", round(mean_after, 4))
print("두 평균의 차이:", round(diff, 4))
````

## 문제 ② Min-Max / Z 정규화 평균 비교

`Wind` 변수에 대해서 **Min-Max 정규화**를 수행한 후 평균값과  
**Z 정규화(표준화)**를 수행한 후 평균값의 차이를 구하시오.

- Min-Max 정규화: $(x - min) / (max - min)$
- Z 정규화: $(x - mean) / std$

> 출력: 두 평균의 차이

````python
wind = airquality['Wind']
minmax = (wind - wind.min()) / (wind.max() - wind.min())
z = (wind - wind.mean()) / wind.std()
diff = abs(minmax.mean() - z.mean())

print("Min-Max 정규화 평균:", round(minmax.mean(), 4))
print("Z 정규화 평균:", round(z.mean(), 4))
print("두 평균의 차이:", round(diff, 4))
````

## 문제 ③ 월별 평균 기온

월별(5월~9월) **평균 기온(`Temp`)**을 구하시오.

> 출력: Month별 Temp 평균

````python
monthly_temp = airquality.groupby('Month')['Temp'].mean()
monthly_temp
````
