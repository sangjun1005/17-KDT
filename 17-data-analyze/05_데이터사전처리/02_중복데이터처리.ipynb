```python
import pandas as pd
```

```python
df = pd.DataFrame({
    'c1':['a','a','b','a','b'],
    'c2':[1,1,1,2,2],
    'c3':[1,1,2,2,2]
})

df
```

```python
df_dup_first = df.duplicated()
df_dup_first
```

```python
df_dup_last = df.duplicated(keep='last')
df_dup_last
```

```python
df_dup_false = df.duplicated(keep=False)
df_dup_false
```

```python
col_dup = df['c2'].duplicated()
col_dup
```

```python
col_dup2 = df.duplicated(subset=['c2'])
col_dup2
```

```python
df2 = df.drop_duplicates()
df2
```
