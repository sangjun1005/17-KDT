```python
import pandas as pd
import seaborn as sns

titanic = sns.load_dataset('titanic')
df = titanic.loc[:,['age','sex','class','fare','survived']]

grouped = df.groupby(['sex','class'],observed=True)

gdf = grouped.agg(['mean',"std"],numeric_only=True)

gdf
```

```python
gdf.index
```

```python
arrays = [
    ['a','a','b','b'],
    [1,2,1,2]
]

multi_index_arrays = pd.MultiIndex.from_arrays(arrays,names=('letter','number'))
multi_index_arrays
```

```python
tuples = [
    ('a',1),
    ('a',2),
    ('b',1),
    ('b',2)
]

multi_index_tuples = pd.MultiIndex.from_tuples(tuples,names=('letter','number'))
multi_index_tuples
```

```python
letters = ['a','b']
numbers = [1,2]

multi_index_prodect = pd.MultiIndex.from_product([letters,numbers],names=('letter','number'))

multi_index_prodect
```
