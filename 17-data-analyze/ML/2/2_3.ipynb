```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
titanic_df = pd.read_csv('./data/train.csv')
titanic_df.head()
```

```python
y_titanic_df = titanic_df['Survived']
X_titanic_df = titanic_df.drop('Survived', axis=1)
X_titanic_df.head()
```

```python
X_titanic_df.info()
```

```python
test_df = pd.read_csv('./data/test.csv')

res = test_df['PassengerId']
test_df.tail()
```

```python
test_df.info()
```

```python
for df in [X_titanic_df, test_df]:
    df['Age'] = df['Age'].fillna(df['Age'].mean())
    df['Cabin'] = df['Cabin'].fillna('N')
    df['Embarked'] = df['Embarked'].fillna('N')
    df['Fare'] = df['Fare'].fillna(0)
    df['Cabin'] = df['Cabin'].str[:1]

for df in [X_titanic_df, test_df]:
    print(df.isnull().sum())
```

```python
drop_col = ['PassengerId', 'Name', 'Ticket']
X_titanic_df = X_titanic_df.drop(columns = drop_col)
test_df = test_df.drop(columns = drop_col)
```

```python
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, OrdinalEncoder
```

```python
oe = OrdinalEncoder(handle_unknown = 'use_encoded_value', unknown_value=-1)
X_titanic_df[['Cabin', 'Sex', 'Embarked']] = oe.fit_transform(X_titanic_df[['Cabin', 'Sex', 'Embarked']])
test_df[['Cabin', 'Sex', 'Embarked']] = oe.transform(test_df[['Cabin', 'Sex', 'Embarked']])
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_titanic_df, y_titanic_df, test_size=0.2, random_state=42
)

X_train, X_valid, y_train, y_valid = train_test_split(
    X_train, y_train, test_size=0.25, random_state=42
)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# 모델 생성

dt_clf = DecisionTreeClassifier(random_state=42)
rf_clf = RandomForestClassifier(random_state=42)
lr_clf = LogisticRegression(solver='liblinear')

# 모델 학습 및 검증

dt_clf.fit(X_train, y_train)
dt_pred = dt_clf.predict(X_valid)
print('DecisionTree Accuracy:', accuracy_score(y_valid, dt_pred))

rf_clf.fit(X_train, y_train)
rf_pred = rf_clf.predict(X_valid)
print('RandomForest Accuracy:', accuracy_score(y_valid, rf_pred))

lr_clf.fit(X_train, y_train)
lr_pred = lr_clf.predict(X_valid)
print('LogisticRegression Accuracy:', accuracy_score(y_valid, lr_pred))
```

```python
dt_test = dt_clf.predict(test_df)
rf_test = rf_clf.predict(test_df)
lr_test = lr_clf.predict(test_df)

result = pd.DataFrame({'PassengerId': res,
                       'DecisionTree': dt_test,
                       'RandomForest': rf_test,
                       ':LogisticRegression': lr_test})

result.head()
```
